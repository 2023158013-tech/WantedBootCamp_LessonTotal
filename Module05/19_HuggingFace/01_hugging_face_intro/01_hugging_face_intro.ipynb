## 01. Hugging Face를 해보는 이유

* 이번 챕터에서는 아래 AI 기능을 구현할 것이다.
- 감정 분석
- 문서 분류
- 문서 요약
- 객체 추출

* 이러한 기능을 직접 AI 모델로 만들고 학습시키는 것은 현실적으로 불가능하다.
* 따라서 이미 학습된 AI 모델을 가져와 서비스에 적용하는 것이 일반적인 개발 방식이다.
* AI 모델들과 DataSet들이 모여진 Github같은 사이트가 "hugging face"라는 플랫폼이다.

Hugging Face를 활용하는 방법

* <과거 방식>
    - 모델 직접 학습
    - 모델 저장
    - 추론할 수 있는 코드 작성

* <Hugging Face 방식>
    - 원하는 모델 검색
    - 모델 다운로드
    - Python 코드 작성
