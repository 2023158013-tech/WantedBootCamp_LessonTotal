## 01. Python 변수
* 변수란? 값을 담을 수 있는 공간이다.
* Java와는 다르게 Python은 변수의 타입을 선언 시에 지정하지 않는다.
* Python은 런타임 시점에서 값을 보고 타입을 판단하게 된다.

```python
user_question = "python이 뭐야?"

print(user_question)
```

## 02. 변수 이름 작성 규칙
* Python에서는 보통 변수명을 snake_case로 작성한다.
* Java는 cammel_case로 작성한다.
* ex) userName <- java 표현식 | user_name <- python 표현식
* (암묵적인 규칙)

```python
model_name = "stt-ai-model"
max_token = 1000
temperature = 0.7
is_streaming = True

print(model_name)
print(max_token)
print(temperature)
print(is_streaming)
```

## Python의 자료형은 런타임 시 결정된다.
* type() 메서드 함수로 변수의 타입을 확인할 수 있다.

```python
print(type(model_name))
print(type(max_token))
print(type(temperature))
print(type(is_streaming))
```

```python
name = "raccoon"
user_question = "RAG가 뭐야?"

# raccoon 사용자의 질문: RAG가 뭐야? => 이런 식으로 문자열 합치기

print(name + "의 질문: " + user_question)

# 또 다른 문자열 합치는 방법: f-string(+없이도 간편하게 문자열을 합칠 수 있다.)
message = f"{name}의 질문: {user_question}"
print(message)
```
