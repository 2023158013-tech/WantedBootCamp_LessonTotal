## 05. List
* Python List는 Java 진영에서의 ArrayList와 같은 구조
* List가 왜 필요할까?
users = [
    {user01, pass01}, # 객체(dictionary 타입) -> 이게 여러 개 모이면 list
    {user02, pass02},
    {user03, pass03},
]
* 하나의 데이터만을 다루는 경우보다 여러 개의 데이터를 다루는 경우가 훨씬 많다.

````python
# Python List 기본 형태(배열)

# Java []: 배열
# 배열의 특징: 크기가 정해져야 함, 순서가 있음, 지료형이 같아야 함.

# Python []: List
# Java의 배열과 다른 가장 큰 차이점: 자료형은 상관이 없다.
messages = [
    "안녕",
    "HI",
    "raccoon"
]

print(messages)
print(type(messages))
````

````python
messages = [
    "안녕",
    "HI",
    "raccoon"
]

# 데이터에 접근하는 방법
# 배열의 특성을 떠올려보자. Index 번호 체계에 의한 순서가 존재한다!!

print(messages[0]) # 변수명[index]

# 데이터를 추가하는 방법
# ex) List<UserDTO> users = new ArrayList<>;
# users에 4번 회원을 추가하려면? users.add (java 방법)
# python 방법: append() 메서드가 Java 진영에서의 add()와 동일한 역할을 하게 된다.

messages.append("koala")

print(messages)

# 데이터를 수정하는 방법
messages[3] = "tiger"

print(messages)

# 데이터를 삭제하는 방법
# remove(), pop()

messages.remove("tiger") # 값로 삭제
messages.pop(0) # 인덱스로 삭제

print(messages)

# 배열의 크기를 알아보는 메서드(Java에서는 .size())
# 배열 사용의 장점? 특징과 연관(순서가 있다.) -> 반복을 돌릴 수 있다.
# java -> size()
# python -> len() (length의 약자)

print(len(messages))

# Python에서의 반복문 : Python에서의 반복문은 Java에서 향상된 for문과 비슷한 구조이다.
# ```java
#//향상된 for문(for-each)
#for(String message: messages) {
#    System.out.println(message);
#}
#``` 같은 의미
messages2 = ["java", "js", "python", "c#", "c"]

for message in messages2:
    print(message)
````

````python
# 지금까지 배운 개념 누적 적용
class AIService:
    
    def answer(self, question: str) -> dict:
        return {
            "question" : question,
            "answer" : f"{question}에 대한 답변입니다."
        } # dictionary라서 중괄호로 반환
        
ai_service = AIService()

questions = [
    "Python이 뭐야?",
    "점심 뭐 먹지?",
    "FastAPI가 뭐야? 도대체?"
]

responses = []

for question in questions:
    response = ai_service.answer(question)
    responses.append(response)
    
print(responses)
````
