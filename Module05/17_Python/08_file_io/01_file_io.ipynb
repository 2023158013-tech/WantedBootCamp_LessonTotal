## 08. File Input / Output

* File 입출력은 AI 서버와 RAG에서 굉장히 빈번하게 사용한다.
* AI Model은 Data가 많으면 많을수록 정확도 있는 답변을 줄 수 있다.
* 또한 회사에서는 Data들이 다양한 형태(.hwp, .csv, .txt, .pdf 등)로 저장되어 있으며 해당 파일들을 읽어 Model을 학습시키게 된다.

* 자주 쓰이는 예: 공지사항 파일 읽기, 회사 내규 파일 읽기, DB상에 저장하지 않는 csv, pdf, txt, hwp 등

---

* RAG
* 대략적인 흐름: 회사 내규 파일 읽기 -> 문서 내용을 잘게 나누기(Chunk 단위) -> Embedding 생성 -> Vector DB 저장 
* -> 사용자의 질문과 관련된 문서 Vector DB에서 검색 -> LLM 답변 생성

```python
# 파일 I/O 관련 기본 문법
# 파일을 읽을 때는 open() 함수를 사용한다.
# with 문법: Java에서의 try-with-resource와 동일한 역할을 한다.
# 파일 사용이 끝나게 되면 자동으로 Stream(파일을 읽기 위한 통로 자원) 등 자원들을 종료시켜준다.

# open("파일명", "모드", "인코딩 방식")
with open("sample_question.txt", "r", encoding="UTF-8") as file:
    lines = file.readlines()
    
print(lines)

# ex) public String test() { return "안녕"; }
# test(); //호출(메서드의 반환타입이 String)
# 리스트로 읽어오는 이유는 readlines() 함수의 타입이 리스트로 설정되어 있기 때문.
```

```python
# 주의사항: "w" 모드는 기존의 내용을 덮어쓴다.
answer = "내일의 날씨는 습도가 1%, 최고 기온은 33도입니다."

# 쓰기
with open("ai_answer.txt", "w", encoding="UTF-8") as file:
    file.write(answer)
    
print("저장 완료!")
```

```python
# "a" 모드: append의 약자로서 기존 파일에 내용을 추가(누적)할 때 사용한다.
answer = "내일 모레의 날씨는 습도가 20%, 최고 기온은 30도입니다."

# 쓰기
with open("ai_answer.txt", "a", encoding="UTF-8") as file:
    file.write(answer)
    # append는 같은 행에 주가가 된다.
    # 만약 한 줄 띄우기를 하고자 하면 \n 개행 문자를 포함해야 한다.
    # ex) "\n오늘의 날씨는 비가 옵니다."
print("저장 완료!")
```

## 지금까지 배운 내용을 누적해서 적용!

```python
# dictionary 복습

with open("sample_question.txt", "r", encoding="utf-8") as file:
    question = file.read().strip() # 공백 삭제
    
# JAVA에서의 Map<String, Object>과 같은 역할을 하게 되며, JSON 구조와 동일하다. {key-value}
response = {
    "question": question,
    "answer": f"{question}에 대한 예시 답변"
}

print(response)
```

```python
# List

# [] <- 자바에서는 배열이지만 파이썬에서는 List이다.
documents = []

with open("rag_documents.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()
    
# Python for문은 Java에서의 향상된 for문과 문법상 유사하다.
# for {꺼낸 값 담을 변수명} in {반복할 변수명}
for line in lines:
    documents.append(line.strip())
    
print(documents)
```

```python
class RAGService:
    def answer(self, question: str, documents: list) -> dict:
        
        context = "\n".join(documents)
        
        return {
            "question": question,
            "context": context,
            "answer": "제공된 문서를 바탕으로 LLM이 답변을 생성했습니다."
        }

rag_service = RAGService() # 인스턴스 대입

# Spring Server에 보내줄 AI 챗봇 결과 JSON
response = rag_service.answer("RAG가 도대체 무엇입니까?", documents)

print(response)
```
