## 04. Dictionary
* Dictionary는 Java에서의 Map과 비슷한 역할을 한다.
* Map<String, Object> 형태라고 생각하는 것이 좋다.
* Json: { name : "raccoon" }
* 🚨 Key - Value 형태의 자료구조이다!!!!!

```python
user = {
    "name" : "raccoon",
    "age" : 20,
    "course" : "AI 활용 Backend 과정"
} # JSON과 똑같이 key-value, 중괄호 사용

print(user)
print(type(user)) # dict 타입 = dictionary의 약자
```

```python
# <dictionary 타입의 값을 꺼내오는 방법>
# dictionary에 들어있는 값 조회(값 꺼내는 것이 중요)
print(user["name"]) # ["key"] -> value
print(user.get("age")) # get("key") -> value (getter 메서드)
```

```python
# <dictionary 타입의 값을 수정하는 방법>
user["name"] = "koala"

# <dictionary 타입의 값을 삭제하는 방법>
del user["age"]

print(user)
```

```python
# Class와 함께 Dictionary 자료형 활용해보기
# 1~4 총 집합

# AI 응답 관련 Service Class
class AIService:
    
    def answer(self, request : dict) -> dict :
        return {
            "question" : request["question"],
            "model" : request["model"],
            "answer" : f"{request.get("question")}에 대한 답변!"
        }
        
# AI 응답 객체 생성
service = AIService()

# AI 응답 요청 해보기
response = service.answer({ # 여기에 들어가는 데이터는 스프링에서 보내줄 것, AI 서버에서는 모델을 가져와서 응답하는 것
    "question" : "FastAPI가 도대체 뭐야?",
    "model" : "gemini-slim"
})

print(response)
```
