## 07. Exception
* Python에서의 예외 처리
* AI 관련 서버에서 자주 발생하는 예외 케이스
- 질문이 비어있음 (Spring 백엔드에서 충분한 Filter를 해주면 OK)
- 존재하지 않거나 버전이 만료된 모델
- API 호출 실패 (EndPoint가 잘못된 것보단 API Key 만료, 초과 사용이 원인인 경우가 큼)
- FileNotFound (RAG 시스템 구축 시 pdf, excel, sheet, csv 등의 파일을 활용(I/O)하여 답변 퀄리티를 높이게 되는데 이때 file 관련 에러가 빈번히 발생)

* 위 상황들을 예외 처리하지 않으면 프로그램이 비정상적으로 종료됨.


```python
# Exception 기본 문법
# try - catch (Java)
# try - except (Python)

try:
    number = 10 / 0
except Exception as e:
    print(f"{e}예외 발생")

print("프로그램이 계속 동작하나?")
```

```python
# 명시적으로 Exception 발생 시키기

question = ""

if question == "":
    # throw new IllegalArgumentException("숫자가 아닙니다")
    # raise는 우리가 직접 예외를 발생시킬 수 있다.
    raise ValueError("질문은 비어있을 수 없습니다.")
    
```

```python
# 우리가 지금까지 배운 개념을 총 동원해서 예외 상황에 적용

class AIService:
    
    def answer(self, question :str) -> dict:
        
        # strip(): "" != " " (space) => 띄워쓰기를 지워준다.
        if question.strip() == "" :
            raise ValueError("질문은 비어있을 수 없다")
        return {
            "answer" : f"{question}에 대한 AI의 답변"
        }
        
service = AIService() # AIService 인스턴스 생성

# Spring에서 넘어온 질문
question = "    "

try:
    response = service.answer(question)
    print(response)
except ValueError as e:
    print("AI Service Error 발생")
    print(f"발생한 Error: {e}")

print("프로그램은 계속 동작합니다.")
```
