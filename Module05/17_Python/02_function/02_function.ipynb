## 02. Function
* Java에서 Method의 개념이 Python과 JavaScript에서는 Function의 개념과 동일하다.
* Java에서의 Method는 class 내부에서만 작성이 가능하다.
* js, py에서는 function도 자체적으로 객체이기 때문에 별도의 클래스가 없어도 작성 가능하다.

````python
# 함수 없이 같은 코드를 반복
user_question_1 = "LangChain이 뭐야?"
answer_1 = user_question_1 + "에 대한 답변"

print(answer_1)

user_question_2 = "FastAPI가 뭐야?"
answer_2 = user_question_2 + "에 대한 답변"

print(answer_2)
````

## function의 기본 문법
* 함수를 선언할 때는 def 키워드를 활용해서 함수를 만든다.

```java
public void hello() {
    System.out.println("hello!!!");
}
```

```python
def hello():
    print("hello!!!")
```

````python
def say_hello():
    print("hello py")
    
# Python에서는 {구역}이 들여쓰기(Tab)으로 결정된다.
# 따라서 들여쓰기를 잘못하게 된다면 영역을 이탈 or 침범하게 되므로 주의해야 한다.
say_hello() # 함수 호출(함수명())
````

## 매개변수가 있는 Function

````python
def create_answer(question):
    return question + "에 대한 답변!!"

result = create_answer("오늘 점심 메뉴가 뭐야?")

print(result)
````

````python
# Type-Hint: 타입에 대한 힌트
# Python은 런타임 시점에 타입이 결정되기 때문에 여러 개발자들이 협업 시에 해당 변수가 int, str인지 알려주기 위한 용도로 사용된다.
def create_answer2(question : str) -> str :
    return question + "에 대한 답변!!"

result2 = create_answer2("오늘 점심 메뉴가 뭐야?")

print(result2)
````

````python
# 기본값이 존재하는 함수
# 해당 함수의 매개변수에 아무런 값이 넘어지 않더라도 설정한 기본값을 출력할 수 있다.
def create_answer3(question : str, model_name : str = "tts-ai-model") -> str :
    return f"{question} + 에 대한 답변: {model_name}"

result3 = create_answer3("룰루룰루ㅜ우루울ㅇ~~")
result4 = create_answer3("룰루룰루ㅜ우루울ㅇ~~", "stt-ai-model")

print(result3)
print(result4)
````

````python
# 조건문
def valid_question(question : str) -> bool :
    # Java = Null, Py = None
    if question is None:
        return False
    return True

print(valid_question(None))
print(valid_question("HI"))
print(valid_question("")) # 빈 문자열은 문자열(None이 아님)
````
