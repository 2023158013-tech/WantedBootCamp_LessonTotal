## 04. Router

* FastAPI의 APIRouter를 활용해서 API 관련 코드를 분리하는 방법
* 문제 상황: 
    * main.py 파일에 모든 API 및 DTO를 작성해두었다.
    * 프로젝트의 규모 및 기능이 추가될수록 main.py가 복잡해진다.

* 필요 사항:
    * Route의 개념이 필요하게 된다.
    * Spring의 Controller == FastAPI의 Router 동일한 역할
    * APIRouter를 사용해서 main.py 파일에 Router를 등록한다.

* <총 구조>
* main.py
    * FastAPI 객체 생성
    * Router 등록

    * ChatBot 관련 기능 API 모아 작성
* rag_router.py
    * RAG 관련 기능 API 모아 작성

* __init__.py 파일을 각각의 폴더에 추가했다.
* Python 3.X 버전에서는 __init__.py 파일이 없다면 단순한 폴더로만 인식한다.
* __init__.py 파일이 있다면 단순 폴더가 아닌 Python Package로 인식하기 때문에 import 시 Error가 발생하지 않는다.
