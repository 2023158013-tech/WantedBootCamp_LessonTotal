## 00. FastAPI란?
* Python기반의 API 서버를 만들 수 있는 웹 Framework
* Spring Framework와 동일한 역할을 한다.
---
FastAPI 특징
* 빠르게 API를 개발할 수 있다.(경량)
* 자동 Swagger 문서가 설정된다.(Java에서는 의존성을 주입해야 했음.)
* Pydantic 기반의 요청/응답 검증(spring의 validation과 비슷.)
* 비동기 처리 지원
* Python 생태계이기 때문에 AI/ML/DL 관련 서버로 활용하기 좋음.
* 즉, 라이브러리/래퍼런스가 풍부하다.

## 01. 전체 흐름 구조
Client(사용자) -요청-> Next.js(frontend server) -버튼 클릭-> Spring(backend server)
-AI 요청-> FastAPI ai server(≒backend server) -> PromptService(프롬프트 조립 후 넘김) -> ChatBotService(조립된 프롬프트와 함께 모델 호출)
-> AI Model -> 답변 시작
